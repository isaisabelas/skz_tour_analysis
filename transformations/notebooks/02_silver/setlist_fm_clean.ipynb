```python
from pyspark.sql import functions as F
```

```python
df = spark.table("skztour.01_bronze.raw_setlist_fm")
display(df)
```

Transformações: 
<br>
- Converte o campo event_date de string para tipo date
- Limpar whitespace dos campos de texto
- Extrair country do campo venue_city
- Adicionar coluna source
- Criar mapeamento cidade/país


```python
display(df)
```

```python
# Converte o campo event_date de string para tipo date
# Formato original: "dd-MM-yyyy" (ex: "06-06-2026")
df = df.withColumn("event_date", F.try_to_date(df["event_date"], "dd-MM-yyyy"))
```

```python

df_clean = df \
    .withColumn("source", F.lit("setlist_fm")) \
    .withColumn("artist_name", F.trim(F.col("artist_name"))) \
    .withColumn("venue_name", F.trim(F.col("venue_name"))) \
    .withColumn("venue_city", F.trim(F.col("venue_city"))) \
    .withColumn("event_year", F.year("event_date")) \
    .select(
        F.col("setlist_id").alias("show_id"),
        "source",
        "event_date",
        "artist_name",
        "venue_name",
        "venue_city",
        F.lit(None).cast("string").alias("country"),  # Adicionar depois
        F.lit(None).cast("string").alias("attendance"),
        F.col("url").alias("source_url")
    )

print(f"Total shows: {df_clean.count()}")
```

```python
display(df_clean)
```

```python
# Primeiro, vamos ver quais cidades temos
print("Cidades únicas no dataset:")
df_clean.select("venue_city").distinct().orderBy("venue_city").show(50, truncate=False)
```

```python
# Criar dicionário de mapeamento cidade → país
city_to_country = {
    "Chiba": "Japan",
    "Goyang-si": "South Korea",
    "Hong Kong": "Hong Kong",
    "Incheon": "South Korea",
    "Kaohsiung City": "Taiwan",
    "Nanterre": "France",
    "Queens": "United States",
    "Seoul": "South Korea",
    "Taipei": "Taiwan",
    "Tokyo": "Japan"
}

# Criar função de mapeamento
def map_city_to_country(city):
    return city_to_country.get(city, None)  # Retorna None se cidade não encontrada

# Registrar como UDF
from pyspark.sql.types import StringType
map_country_udf = F.udf(map_city_to_country, StringType())

# Aplicar ao DataFrame
df_clean_with_country = df_clean.withColumn("country", map_country_udf(F.col("venue_city")))

print("\nVerificação do mapeamento:")
df_clean_with_country.select("venue_city", "country").distinct().orderBy("venue_city").show(truncate=False)
```

```python
display(df_clean_with_country)
```

```python
# Verificar resultado final com schema padronizado
print("Schema final:")
df_clean_with_country.printSchema()
```

```python
# Salvar na camada silver (com país mapeado)
df_clean_with_country.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("skztour.02_silver.clean_setlist_fm")

print("✅ Tabela clean_setlist_fm salva com sucesso!")
print(f"Total shows: {df_clean_with_country.count()}")
```
